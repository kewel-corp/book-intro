## 13.4　関数 ― 処理をまとめる

````python
def dice_score(pred, target):
    """予測マスクと正解マスクのDice係数を計算する"""
    intersection = (pred * target).sum()
    return 2.0 * intersection / (pred.sum() + target.sum() + 1e-8)

# pred・target は予測マスクと正解マスク（NumPy配列。第14章で学びます）。呼び出しの形は次のとおり：
# score = dice_score(pred, target)
# print(f"Dice = {score:.3f}")
````

## 13.5　クラスとオブジェクト ― PyTorchを読むための最小知識

````python
class Case:
    def __init__(self, case_id, tumor_mm):   # 初期化：オブジェクトを作るとき呼ばれる
        self.case_id = case_id               # self に値を持たせる（属性）
        self.tumor_mm = tumor_mm

    def is_small(self):                      # メソッド：このオブジェクトの処理
        return self.tumor_mm <= 10

case = Case("pancreas_001", 8.5)   # オブジェクトを作る（設計図から実体へ）
print(case.case_id)                # 属性を参照 → "pancreas_001"
print(case.is_small())             # メソッドを呼ぶ → True
````
