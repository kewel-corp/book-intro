## 7.2　GitHubとは ― コードを置くクラウド

```bashgit clone https://github.com/kewel-corp/book-intro.git   # 本シリーズのコードを手元にコピー# （ファイルを編集したあと）git add .                          # 変更を記録の対象にするgit commit -m "OCT分類のコードを追加"   # 変更を履歴として刻む（-mはメッセージ）git push                           # GitHubへ送る（共有・バックアップ）git pull                           # 他人の変更を取り込む```
